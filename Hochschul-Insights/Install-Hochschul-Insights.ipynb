# Hochschul-Insights — Fabric Notebook Installer

Deploys the full **Hochschul-Insights** demo into the **current Fabric workspace**:

- Lakehouse `WebinarLakehouse` (schemas enabled, schema `Genesis`)
- Loader Notebook (GENESIS Loader **or** Snapshot Loader depending on mode) + Dimensions notebook
- Data Pipeline (live mode only)
- Direct Lake Semantic Model `Hochschule`
- Report `Webinar Hochschule` (8 pages, IBCS + Azure Map)
- DataAgent `Hochschul-Stats-Agent` (best-effort)

Everything lands in a new workspace folder named `Hochschul-Insights`.

## How to use

1. **Import this notebook** into your target Fabric workspace.
2. Choose a mode:
   - **Snapshot mode (no token, fastest):** leave `GENESIS_TOKEN` empty. The installer loads bundled CSV snapshots from the public GitHub repo so the report works immediately.
   - **Live mode:** paste a free DESTATIS GENESIS token. Register at https://www-genesis.destatis.de/ and copy your **username** token from your profile. The installer creates and runs a real pipeline.
3. Run all cells.

No `az login`, no `pip install`, no env variables needed — the notebook authenticates itself via `notebookutils.credentials.getToken("pbi")` and auto-detects the current workspace.

Repository: https://github.com/KornAlexander/Fabric-Demos

Leave `GENESIS_TOKEN` **empty for snapshot mode** (bundled CSVs, fast, no signup).

## Run the installer

Downloads the latest `Install-Hochschul-Insights.py` from GitHub and runs it with the parameters above.

```python
import urllib.request, types

INSTALLER_URL = "https://raw.githubusercontent.com/KornAlexander/Fabric-Demos/main/Hochschul-Insights/Install-Hochschul-Insights.py"

src = urllib.request.urlopen(INSTALLER_URL).read().decode("utf-8")
mod = types.ModuleType("hi_install")
exec(compile(src, INSTALLER_URL, "exec"), mod.__dict__)

result = mod.install(workspace_id=WORKSPACE_ID, genesis_token=GENESIS_TOKEN, run_pipeline=True)
result
```
