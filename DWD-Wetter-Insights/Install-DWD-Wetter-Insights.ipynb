# DWD-Wetter-Insights — Fabric Notebook Installer

Deploys the full **DWD-Wetter-Insights** demo into the **current Fabric workspace**:

- Lakehouse `DemoLakehouse` (schemas enabled)
- 4 Notebooks: Loader (dwdown), Forecast Loader (Mosmix), Refresh SM, Notify
- Direct Lake Semantic Model `Wetter-Insights`
- Report `DWD Wetter-Insights` (4 pages: Home, Übersicht, Stationen, Vorhersage)
- Data Pipeline `DWD-Wetter-Insights Daily` (Load → Refresh → Notify)

Everything lands in a new workspace folder named `DWD-Wetter-Insights`.

## How to use

1. **Import this notebook** into your target Fabric workspace.
2. Run all cells.

No `az login`, no `pip install`, no env variables, no API tokens — DWD Climate Data Center is fully open.
The notebook authenticates itself via `notebookutils.credentials.getToken("pbi")` and auto-detects the current workspace.

Repository: https://github.com/KornAlexander/Fabric-Demos

## Parameters (optional)

```python
# Optional: override the target workspace (default = the workspace this notebook runs in)
WORKSPACE_ID = None  # e.g. "da2e15a8-c06d-4da0-ad10-c68aba63e564"

# Optional: lakehouse name (default = DemoLakehouse). If a lakehouse with this name
# already exists in the workspace, it will be reused instead of recreated.
LAKEHOUSE_NAME = "DemoLakehouse"
```

## Run the installer

Downloads the latest `Install-DWD-Wetter-Insights.py` from GitHub and runs it with the parameters above.

```python
import urllib.request, types

INSTALLER_URL = "https://raw.githubusercontent.com/KornAlexander/Fabric-Demos/main/DWD-Wetter-Insights/Install-DWD-Wetter-Insights.py"

src = urllib.request.urlopen(INSTALLER_URL).read().decode("utf-8")
mod = types.ModuleType("dwd_install")
exec(compile(src, INSTALLER_URL, "exec"), mod.__dict__)

result = mod.install(workspace_id=WORKSPACE_ID, lakehouse_name=LAKEHOUSE_NAME, run_pipeline=True, wait_for_pipeline=True)
result
```

## Next steps

1. Open the workspace → folder `DWD-Wetter-Insights`.
2. Open **`DWD Wetter-Insights`** report — Direct Lake lights up automatically after the first load.
3. Schedule the **`DWD-Wetter-Insights Daily`** pipeline daily at 06:00 (DWD publishes overnight).
